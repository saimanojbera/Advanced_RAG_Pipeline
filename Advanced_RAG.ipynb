Getting all the catergories and there links from base url: https://www.renesas.com/en/applications/automotive

```python
import asyncio
import json
from urllib.parse import urljoin
from crawl4ai import AsyncWebCrawler, BrowserConfig, CrawlerRunConfig, CacheMode
from crawl4ai.extraction_strategy import JsonCssExtractionStrategy

async def main():
    base_url = "https://www.renesas.com/en/applications/automotive"

    # Enable JavaScript for dynamic content
    browser_config = BrowserConfig(headless=True, java_script_enabled=True)

    # Updated schema with correct div structure
    schema = {
        "name": "All_Links_from_base_url",
        "baseSelector": "div.rgrid.rgrid--col3 div.rcard.rcard--has-gradient",
        "fields": [
            {"name": "title", "selector": "a", "type": "text"},  # Extract title
            {"name": "link", "selector": "a", "type": "attribute", "attribute": "href"}  # Extract href
        ]
    }

    async with AsyncWebCrawler(config=browser_config) as crawler:
        print("Extracting all categories links...")
        result = await crawler.arun(
            url=base_url,
            config=CrawlerRunConfig(
                cache_mode=CacheMode.BYPASS,
                extraction_strategy=JsonCssExtractionStrategy(schema),
                wait_for="div.rcard.rcard--has-gradient",

            )
        )

        # Debugging: Print extracted data
        data = json.loads(result.extracted_content)
        print("Extracted Data:", json.dumps(data, indent=2))


if __name__ == "__main__":
    asyncio.run(main())
```

#Lets iterate with each categorie dynamically keeping base url : https://www.renesas.com/en/applications/automotive  to get the winning models and descriptions in to a Json and also converting the diagrams (SVG) in to Png format for clip embedding.

```python
import asyncio
import json
import os
import cairosvg
from urllib.parse import urljoin
from crawl4ai import AsyncWebCrawler, BrowserConfig, CrawlerRunConfig, CacheMode
from crawl4ai.extraction_strategy import JsonCssExtractionStrategy

async def extract_data():
    base_url = "https://www.renesas.com/en/applications/automotive"

    # Enable JavaScript for dynamic content
    browser_config = BrowserConfig(headless=True, java_script_enabled=True)

    async with AsyncWebCrawler(config=browser_config) as crawler:
        ### **Step 1: Extract All Category Links from Base URL** ###
        print(f" Extracting all category links from {base_url}...")

        schema_categories = {
            "name": "All_Links",
            "baseSelector": "div.rgrid.rgrid--col3 div.rcard.rcard--has-gradient",
            "fields": [
                {"name": "link", "selector": "a", "type": "attribute", "attribute": "href"}
            ]
        }

        category_result = await crawler.arun(
            url=base_url,
            config=CrawlerRunConfig(
                cache_mode=CacheMode.BYPASS,
                extraction_strategy=JsonCssExtractionStrategy(schema_categories),
                wait_for="div.rcard.rcard--has-gradient",

            )
        )

        if not category_result or not category_result.extracted_content:
            print("No categories found - Exiting!")
            return

        category_data = json.loads(category_result.extracted_content)
        categories = [{"link": urljoin(base_url, item["link"])} for item in category_data if "link" in item]

        if not categories:
            print("No categories extracted - Exiting!")
            return

        print(f" Found {len(categories)} categories.")

        final_extracted_data = []

        ### **Step 2: Iterate Over Each Category & Extract Winning Model Links** ###
        for category in categories:
            print(f"🔍 Extracting winning models from {category['link']}...")

            schema_winning_models = {
                "name": "Winning_Model_Links",
                "baseSelector": "ul.application-category-list__group-items li, div.application-category-list",
                "fields": [
                    {"name": "title", "selector": "a", "type": "text"},
                    {"name": "link", "selector": "a", "type": "attribute", "attribute": "href"}
                ]
            }

            try:
                final_links_result = await crawler.arun(
                    url=category["link"],
                    config=CrawlerRunConfig(
                        cache_mode=CacheMode.BYPASS,
                        extraction_strategy=JsonCssExtractionStrategy(schema_winning_models),
                        wait_for="div.application-category-list, main#main-content",

                    )
                )

                if not final_links_result or not final_links_result.extracted_content:
                    print(f" No response for {category['link']} - Skipping!")
                    continue

                final_links_data = json.loads(final_links_result.extracted_content)
                winning_links = [
                    {"title": item["title"].strip(), "link": urljoin(category["link"], item["link"])}
                    for item in final_links_data if "link" in item and item["title"].strip()
                ]

                if not winning_links:
                    print(f" No winning models found for {category['link']}!")
                    continue

                ### **Step 3: Extract Data from Each Winning Model Page** ###
                for entry in winning_links:
                    schema_details = {
                        "name": "Winning_Model_Details",
                        "baseSelector": "div.wysiwyg.field--name-body",
                        "fields": [
                            {"name": "description", "selector": "p", "type": "text"},
                            {"name": "system_benefits", "selector": "ul", "type": "html"}
                        ]
                    }

                    schema_diagram = {
                        "name": "Winning_Model_Diagram",
                        "baseSelector": "div.field--name-field-media-image",
                        "fields": [
                            {"name": "diagram_svg", "selector": "svg", "type": "html"}
                        ]
                    }

                    # Extract details
                    result_details = await crawler.arun(
                        url=entry["link"],
                        config=CrawlerRunConfig(
                            cache_mode=CacheMode.BYPASS,
                            extraction_strategy=JsonCssExtractionStrategy(schema_details),
                            wait_for="div.wysiwyg.field--name-body, main#main-content",

                        )
                    )

                    # Extract diagram
                    result_diagram = await crawler.arun(
                        url=entry["link"],
                        config=CrawlerRunConfig(
                            cache_mode=CacheMode.BYPASS,
                            extraction_strategy=JsonCssExtractionStrategy(schema_diagram),
                            wait_for="div.field--name-field-media-image, main#main-content",

                        )
                    )

                    extracted_details = json.loads(result_details.extracted_content) if result_details and result_details.extracted_content else []
                    extracted_diagram = json.loads(result_diagram.extracted_content) if result_diagram and result_diagram.extracted_content else []

                    # Convert SVG to PNG
                    image_path = None
                    if extracted_diagram and "diagram_svg" in extracted_diagram[0]:
                        output_dir = "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/diagrams"
                        os.makedirs(output_dir, exist_ok=True)
                        png_path = os.path.join(output_dir, f"{entry['title'].replace(' ', '_')}.png")
                        try:
                            cairosvg.svg2png(bytestring=extracted_diagram[0]["diagram_svg"].encode("utf-8"), write_to=png_path)
                            image_path = png_path
                        except Exception as e:
                            print(f" Error converting SVG to PNG: {e}")

                    final_extracted_data.append({
                        "category": category["link"],
                        "model_name": entry["title"],
                        "url": entry["link"],
                        "description": extracted_details[0]["description"] if extracted_details and "description" in extracted_details[0] else "",
                        "system_benefits": extracted_details[0]["system_benefits"] if extracted_details and "system_benefits" in extracted_details[0] else "",
                        "image_path": image_path  # Path to PNG diagram
                    })

            except Exception as e:
                print(f" Error while processing {category['link']}: {str(e)}")
                continue

        # Save as JSON
        json_output_path = "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/winning_models.json"
        with open(json_output_path, "w") as json_file:
            json.dump(final_extracted_data, json_file, indent=4)

        print(f"Data saved as JSON: {json_output_path}")

if __name__ == "__main__":
    asyncio.run(extract_data())
```
