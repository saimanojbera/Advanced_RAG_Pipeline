```python
import asyncio
import json
import os
import re
import cairosvg
from urllib.parse import urljoin
from crawl4ai import AsyncWebCrawler, BrowserConfig, CrawlerRunConfig, CacheMode
from crawl4ai.extraction_strategy import JsonCssExtractionStrategy

async def extract_data():
    base_url = "https://www.renesas.com/en/applications/automotive"
    output_dir = "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/New_Final_winning_models_data"  # Temporary storage
    diagrams_dir = os.path.join(output_dir, "diagrams")

    # Ensure directories exist
    os.makedirs(output_dir, exist_ok=True)
    os.makedirs(diagrams_dir, exist_ok=True)

    browser_config = BrowserConfig(headless=True, java_script_enabled=True)

    async with AsyncWebCrawler(config=browser_config) as crawler:
        print(f"Extracting all category links from {base_url}...")

        schema_categories = {
            "name": "All_Links",
            "baseSelector": "div.rgrid.rgrid--col3 div.rcard.rcard--has-gradient",
            "fields": [
                {"name": "link", "selector": "a", "type": "attribute", "attribute": "href"}
            ]
        }

        try:
            category_result = await crawler.arun(
                url=base_url,
                config=CrawlerRunConfig(
                    cache_mode=CacheMode.BYPASS,
                    extraction_strategy=JsonCssExtractionStrategy(schema_categories),
                    wait_for="div.rcard.rcard--has-gradient",
                )
            )

            if not category_result or not category_result.extracted_content:
                print("No categories found - Exiting!")
                return

            category_data = json.loads(category_result.extracted_content)
            categories = [{"link": urljoin(base_url, item.get("link", ""))} for item in category_data if "link" in item]

            if not categories:
                print("No categories extracted - Exiting!")
                return

            print(f"Found {len(categories)} categories.")

        except Exception as e:
            print(f"Error extracting categories: {str(e)}")
            return

        final_extracted_data = []

        for category in categories:
            print(f"🔍 Extracting winning models from {category['link']}...")

            schema_winning_models = {
                "name": "Winning_Model_Links",
                "baseSelector": "ul.application-category-list__group-items li, div.application-category-list",
                "fields": [
                    {"name": "title", "selector": "a", "type": "text"},
                    {"name": "link", "selector": "a", "type": "attribute", "attribute": "href"}
                ]
            }

            try:
                final_links_result = await crawler.arun(
                    url=category["link"],
                    config=CrawlerRunConfig(
                        cache_mode=CacheMode.BYPASS,
                        extraction_strategy=JsonCssExtractionStrategy(schema_winning_models),
                        wait_for="div.application-category-list, main#main-content",
                    )
                )

                if not final_links_result or not final_links_result.extracted_content:
                    print(f"No response for {category['link']} - Skipping!")
                    continue

                final_links_data = json.loads(final_links_result.extracted_content)
                winning_links = [
                    {"title": item.get("title", "").strip(), "link": urljoin(category["link"], item.get("link", ""))}
                    for item in final_links_data if "link" in item and item.get("title", "").strip()
                ]

                if not winning_links:
                    print(f"No winning models found for {category['link']}!")
                    continue

                for entry in winning_links:
                    schema_details = {
                        "name": "Winning_Model_Details",
                        "baseSelector": "div.wysiwyg.field--name-body",
                        "fields": [
                            {"name": "description", "selector": "p", "type": "text"},
                            {"name": "system_benefits", "selector": "ul", "type": "html"}
                        ]
                    }

                    schema_diagram = {
                        "name": "Winning_Model_Diagram",
                        "baseSelector": "div.field--name-field-media-image",
                        "fields": [
                            {"name": "diagram_svg", "selector": "svg", "type": "html"}
                        ]
                    }

                    # Extract details
                    try:
                        result_details = await crawler.arun(
                            url=entry["link"],
                            config=CrawlerRunConfig(
                                cache_mode=CacheMode.BYPASS,
                                extraction_strategy=JsonCssExtractionStrategy(schema_details),
                                wait_for="div.wysiwyg.field--name-body, main#main-content",
                            )
                        )

                        result_diagram = await crawler.arun(
                            url=entry["link"],
                            config=CrawlerRunConfig(
                                cache_mode=CacheMode.BYPASS,
                                extraction_strategy=JsonCssExtractionStrategy(schema_diagram),
                                wait_for="div.field--name-field-media-image, main#main-content",
                            )
                        )

                        extracted_details = json.loads(result_details.extracted_content) if result_details and result_details.extracted_content else []
                        extracted_diagram = json.loads(result_diagram.extracted_content) if result_diagram and result_diagram.extracted_content else []

                        # Convert SVG to PNG
                        image_path = None
                        if extracted_diagram and "diagram_svg" in extracted_diagram[0]:
                            # Sanitize filename
                            safe_title = re.sub(r'[<>:&' '-,."/\\|?*]', '_', entry['title'])
                            png_path = os.path.join(diagrams_dir, f"{safe_title.replace(' ', '_')}.png")
                            #f"{entry['title'].replace(' ', '_')}

                            svg_data = extracted_diagram[0].get("diagram_svg", "").strip()

                            # Validate SVG before conversion
                            if svg_data.startswith("<svg"):
                                try:
                                    cairosvg.svg2png(bytestring=svg_data.encode("utf-8"), write_to=png_path, output_width=800)
                                    image_path = png_path
                                    print(f"✅ Successfully converted {entry['title']} to PNG")
                                except Exception as e:
                                    print(f"⚠️ Error converting SVG to PNG for {entry['title']}: {e}")
                            else:
                                print(f"⚠️ Invalid SVG format, skipping {entry['title']}")

                        final_extracted_data.append({
                            "category": category["link"],
                            "model_name": entry["title"],
                            "url": entry["link"],
                            "description": extracted_details[0].get("description", "") if extracted_details else "",
                            "system_benefits": extracted_details[0].get("system_benefits", "") if extracted_details else "",
                            "image_path": image_path  # Path to PNG diagram
                        })

                    except Exception as e:
                        print(f"⚠️ Error processing {entry['link']}: {str(e)}")
                        continue

            except Exception as e:
                print(f"⚠️ Error processing {category['link']}: {str(e)}")
                continue

        # Save JSON output
        json_output_path = os.path.join( "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/New_Final_winning_models_data/new_final_winning_models.json")
        with open(json_output_path, "w") as json_file:
            json.dump(final_extracted_data, json_file, indent=4)

        print(f"✅ Data saved as JSON: {json_output_path}")

if __name__ == "__main__":
    asyncio.run(extract_data())
```

```python
import json
import re
from bs4 import BeautifulSoup

# Load the JSON file
file_path = "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/New_Final_winning_models_data/new_final_winning_models.json"
with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

# Process each entry in the JSON
for entry in data:
    model_name = entry.get("model_name", "")

    # Modify description field
    if "description" in entry:
        entry["description"] = f"This is about application {model_name}: {entry['description']}"

    # Modify system_benefits field and remove HTML tags
    if "system_benefits" in entry:
        clean_text = BeautifulSoup(entry["system_benefits"], "html.parser").get_text()
        entry["system_benefits"] = f"The system benefits for application {model_name} are: {clean_text}"
        entry["system_benefits"] = entry["system_benefits"].replace("\n", " ")

    entry["image_description"] = f"This image is a winning combination for application {model_name}."



# Save the updated JSON back to a file
updated_file_path = "/content/drive/MyDrive/Advanced_RAG_Pipeline_On_Website/New_Final_winning_models_data/cleaned_winning_models.json"
with open(updated_file_path, "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4, ensure_ascii=False)

# Provide the updated file path
updated_file_path
```
